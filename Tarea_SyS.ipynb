# 🔍 Análisis de Circuito RLC Serie con Transformada de Laplace

En este cuaderno analizaremos un circuito **RLC serie** sometido a una entrada $ x(t) $ y obtendremos la ecuación de salida $ y(t) = u_C(t) $, el voltaje sobre el capacitor, usando la **Transformada de Laplace** incluyendo **condiciones iniciales generales**.

Se utilizará el álgebra simbólica de `SymPy` para verificar los resultados paso a paso.


---
## Inicializacion
---

```python
import sympy as sym
sym.init_printing()

# Variables simbólicas
s = sym.symbols('s', complex=True)
t, R, L, C = sym.symbols('t R L C', positive=True, real=True)

# Funciones de entrada y salida
X = sym.Function('X')(s)  # entrada X(s)
Y = sym.Function('Y')(s)  # salida Y(s)

# Valores iniciales
y0, dy0, i0 = sym.symbols('y0 dy0 i0', real=True)
```

## 📘 Ecuación Diferencial del Circuito RLC Serie

Aplicamos la Ley de Kirchhoff de Voltajes:

$$
x(t) = u_R(t) + u_L(t) + u_C(t)
$$

Usamos las relaciones:

- $u_R(t) = R i(t)$  
- $u_L(t) = L \frac{di(t)}{dt}$  
- $u_C(t) = y(t)$  
- $i(t) = C \frac{dy(t)}{dt}$

Sustituimos:

$$
x(t) = R C \frac{dy(t)}{dt} + L C \frac{d^2 y(t)}{dt^2} + y(t)
$$

Entonces, la ecuación diferencial es:

$$
L C \frac{d^2 y(t)}{dt^2} + R C \frac{dy(t)}{dt} + y(t) = x(t)
$$

## 🔁 Transformada de Laplace con Condiciones Iniciales

Aplicamos la Transformada de Laplace a la ecuación anterior, considerando:

- $\mathcal{L}\left\{ \frac{dy}{dt} \right\} = s Y(s) - y(0^+)$  
- $\mathcal{L}\left\{ \frac{d^2 y}{dt^2} \right\} = s^2 Y(s) - s y(0^+) - \dot{y}(0^+)$

Reemplazando en la ecuación:

$$
L C \left[s^2 Y(s) - s y(0^+) - \dot{y}(0^+)\right] + R C \left[s Y(s) - y(0^+)\right] + Y(s) = X(s)
$$


---
### Calculo simbolico
---

```python
# Construcción de Y(s) a partir de la ecuación con condiciones iniciales
numerador = X + (R*C + L*C*s)*y0 + L*i0
denominador = L*C*s**2 + R*C*s + 1

Y_expr = numerador / denominador

# Mostrar la expresión final de Y(s)
sym.pprint(sym.Eq(sym.Symbol('Y(s)'), Y_expr))
```

## ✅ Resultado Final

La salida del sistema en el dominio de Laplace es:

$$
Y(s) = \frac{1}{L C s^2 + R C s + 1} X(s)
+ \frac{R C + L C s}{L C s^2 + R C s + 1} y(0^+)
+ \frac{L}{L C s^2 + R C s + 1} i(0^+)
$$

La respuesta total en el dominio del tiempo se obtiene como:

$$
y(t) = y_{\text{ext}}(t) + y_{\text{int}}(t)
$$

donde:

- $y_{\text{ext}}(t) = \mathcal{L}^{-1}\{Y_{\text{ext}}(s)\}$  
- $y_{\text{int}}(t) = \mathcal{L}^{-1}\{Y_{\text{int}}(s)\}$


---
### Comprobacion paso a paso
---

```python
# Definimos la función de salida y su derivada
y = sym.Function('y')(t)
x = sym.Function('x')(t)

# Construimos la ecuación diferencial (tiempo)
ode = sym.Eq(L*C * y.diff(t,2) + R*C * y.diff(t) + y, x)
print("Ecuación diferencial en el dominio del tiempo:")
display(ode)

# Aplicamos transformada de Laplace con condiciones iniciales
# Transformada de la derivada de segundo orden: s²Y(s) - s*y(0) - y'(0)

lhs_laplace = L*C*(s**2 * Y - s*y0 - dy0) + R*C*(s * Y - y0) + Y
rhs_laplace = X

laplace_eq = sym.Eq(lhs_laplace, rhs_laplace)
print("Transformada de Laplace con condiciones iniciales:")
display(laplace_eq)

# Aislamos Y(s)
Y_expr = sym.solve(laplace_eq, Y)[0]
print("Solución para Y(s):")
display(sym.Eq(sym.Symbol('Y(s)'), Y_expr.simplify()))
```
