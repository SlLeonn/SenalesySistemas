##*Punto 1: Sistema MRD y RLC*

```python
%%writefile pages/1_Punto_1_MRD_RLC.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from control import TransferFunction, feedback, bode_plot
from control.matlab import step, impulse, lsim, pole
import pandas as pd

st.set_page_config(page_title="Punto 1 - MRD y RLC", layout="wide")
st.title("🔧 Punto 1: Sistema Masa-Resorte-Amortiguador & Circuito RLC")

# Sidebar de configuración
st.sidebar.header("⚙️ Configuración del Sistema")
tipo_respuesta = st.sidebar.selectbox("Tipo de Respuesta", ["Subamortiguada", "Sobreamortiguada", "Amortiguamiento Crítico", "Inestable"])
wn = st.sidebar.slider("Frecuencia Natural (ωₙ)", 0.1, 20.0, 5.0)

if tipo_respuesta == "Subamortiguada":
    zeta = st.sidebar.slider("ζ", 0.0, 1.0, 0.3)
elif tipo_respuesta == "Sobreamortiguada":
    zeta = st.sidebar.slider("ζ", 1.0, 2.0, 1.5)
elif tipo_respuesta == "Amortiguamiento Crítico":
    zeta = 1.0
elif tipo_respuesta == "Inestable":
    zeta = st.sidebar.slider("ζ", -1.0, 0.0, -0.1)

# ===============================
# 🔩 Parámetros físicos y sistemas
# ===============================

m = 1.0
k = wn**2
b = 2 * zeta * wn * m
num_mec = [1]
den_mec = [m, b, k]
sys_mec = TransferFunction(num_mec, den_mec)

# Sistema RLC equivalente
L = m
C = 1 / k
R = b
num_elec = [R*C, 0]
den_elec = [L*C, R*C, 1]
sys_elec = TransferFunction(num_elec, den_elec)

# Sistemas en lazo cerrado
sys_mec_cl = feedback(sys_mec, 1)
sys_elec_cl = feedback(sys_elec, 1)

# Mostrar parámetros físicos
col1, col2 = st.columns(2)
col1.subheader("🔧 Sistema Mecánico")
col1.write(f"Masa (m): {m:.2f} kg")
col1.write(f"Amortiguador (b): {b:.2f} Ns/m")
col1.write(f"Resorte (k): {k:.2f} N/m")

col2.subheader("🔌 Sistema Eléctrico")
col2.write(f"Inductor (L): {L:.2f} H")
col2.write(f"Capacitor (C): {C:.4f} F")
col2.write(f"Resistor (R): {R:.2f} Ω")

# ==========================
# 📈 Funciones de graficación
# ==========================

def plot_bode(sys, title):
    try:
        fig = plt.figure()
        bode_plot(sys, dB=True, Hz=False, deg=True)
        plt.suptitle(title)
        st.pyplot(fig)
    except Exception as e:
        st.error(f"⚠ Error en Bode: {e}")

def plot_step(sys, title):
    t = np.linspace(0, 2, 1000)
    t, y = step(sys, T=t)
    fig, ax = plt.subplots()
    ax.plot(y, t)
    ax.set_xlim([0, 2])  # 💡 fuerza eje positivo
    ax.set_title(title)
    ax.set_xlabel("Tiempo (s)")
    ax.grid()
    st.pyplot(fig)

def plot_impulse(sys, title):
    t = np.linspace(0, 2, 1000)
    t, y = impulse(sys, T=t)
    fig, ax = plt.subplots(figsize=(8, 4))  # ¡forzamos proporción!
    ax.plot(y, t)
    ax.set_xlim([0, 2])
    ax.set_title(title)
    ax.set_xlabel("Tiempo (s)")
    ax.set_ylabel("Respuesta")
    ax.grid()
    st.pyplot(fig)

def plot_ramp(sys, title):
    t = np.linspace(0, 10, 1000)
    u = t
    _, y, _ = lsim(sys, U=u, T=t)
    fig, ax = plt.subplots()
    ax.plot(t, y, label="Salida")
    ax.plot(t, u, '--', label="Entrada (rampa)")
    ax.set_xlim([0, 10])  # 💡 fuerza eje positivo
    ax.set_title(title)
    ax.set_xlabel("Tiempo (s)")
    ax.legend()
    ax.grid()
    st.pyplot(fig)

def plot_pzmap(sys, title):
    try:
        p = pole(sys)
        plt.figure(figsize=(6, 6))
        plt.scatter(np.real(p), np.imag(p), marker='x', color='red', label='Polos')
        plt.axhline(0, color='black'); plt.axvline(0, color='black')
        plt.title(title)
        plt.grid(); plt.legend()
        st.pyplot(plt)
    except Exception as e:
        st.error(f"⚠ Error en PZMap: {e}")

# =========================
# 📊 Parámetros temporales
# =========================

if tipo_respuesta == "Subamortiguada" and 0 < zeta < 1:
    try:
        wd = wn * np.sqrt(1 - zeta**2)
        tr = np.pi / wd
        tp = np.pi / wd
        Mp = np.exp(-zeta * np.pi / np.sqrt(1 - zeta**2))
        ts = 4 / (zeta * wn)

        st.markdown("## 📋 Parámetros Temporales")
        st.table(pd.DataFrame({
            "Parámetro": ["Tiempo de levantamiento", "Sobreimpulso máximo", "Tiempo al pico", "Tiempo de establecimiento"],
            "Valor": [f"{tr:.3f} s", f"{Mp*100:.2f}%", f"{tp:.3f} s", f"{ts:.3f} s"]
        }))
    except:
        st.warning("⚠ Error calculando parámetros temporales")

# =============================
# 📈 Visualización del sistema
# =============================

st.markdown("## 🟢 Análisis del Sistema en Lazo Abierto")
col1, col2 = st.columns(2)
with col1:
    st.markdown("### Sistema Mecánico")
    plot_bode(sys_mec, "Bode - Mecánico")
    plot_pzmap(sys_mec, "Polos - Mecánico")
    plot_step(sys_mec, "Escalón - Mecánico")

with col2:
    st.markdown("### Sistema Eléctrico")
    plot_bode(sys_elec, "Bode - Eléctrico")
    plot_pzmap(sys_elec, "Polos - Eléctrico")
    plot_step(sys_elec, "Escalón - Eléctrico")

st.markdown("## 🔁 Análisis en Lazo Cerrado")
col3, col4 = st.columns(2)
with col3:
    st.markdown("### Sistema Mecánico (Lazo Cerrado)")
    plot_bode(sys_mec_cl, "Bode - Mecánico (CL)")
    plot_pzmap(sys_mec_cl, "Polos - Mecánico (CL)")
    plot_step(sys_mec_cl, "Escalón - Mecánico (CL)")
    plot_impulse(sys_mec_cl, "Impulso - Mecánico (CL)")
    plot_ramp(sys_mec_cl, "Rampa - Mecánico (CL)")

with col4:
    st.markdown("### Sistema Eléctrico (Lazo Cerrado)")
    plot_bode(sys_elec_cl, "Bode - Eléctrico (CL)")
    plot_pzmap(sys_elec_cl, "Polos - Eléctrico (CL)")
    plot_step(sys_elec_cl, "Escalón - Eléctrico (CL)")
    plot_impulse(sys_elec_cl, "Impulso - Eléctrico (CL)")
    plot_ramp(sys_elec_cl, "Rampa - Eléctrico (CL)")
```

##*Punto 2: SSB-AM*

```python
%%writefile pages/_Punto_2_SSB_AM.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import hilbert, butter, filtfilt, freqz, tf2zpk
from scipy.fft import fft, fftfreq, fftshift
from scipy.io import wavfile
import os, subprocess
import tempfile
import yt_dlp as youtube_dl

st.set_page_config(page_title="📻 Punto 2 - SSB-AM", page_icon="📡", layout="wide")
st.title("📡 Punto 2: Modulación y Demodulación SSB-AM")

st.markdown("""
## 🧠 Fundamentos Teóricos: Modulación y Demodulación SSB-AM

La **modulación en banda lateral única (SSB-AM)** es una técnica eficiente de modulación de amplitud que elimina una de las bandas laterales y, en ocasiones, también la portadora. Esto permite ahorrar ancho de banda y energía.

### 🔁 Modelo Matemático

#### Dominio del tiempo:
La señal SSB modulada puede escribirse como:

$$
s(t) = m(t)\\cos(2\\pi f_c t) - \\hat{m}(t)\\sin(2\\pi f_c t)
$$

donde:

- $ m(t) $: señal mensaje.
- $ f_c $: frecuencia portadora.
- $ \\hat{m}(t) $: transformada de Hilbert de $ m(t) $, que introduce un desfase de -90°.

#### Dominio de la frecuencia:
La Transformada de Fourier de la señal SSB contiene **solo una banda lateral** (inferior o superior) dependiendo del signo en la ecuación anterior.

---

## 🛠️ Estructura del Panel Interactivo

Este panel permite:

1. **Configurar parámetros clave**:
    - Frecuencia de muestreo y portadora.
    - Frecuencia de corte del filtro paso bajo.
    - Enlace de YouTube para análisis de audio real.

2. **Simulación paso a paso** de:
    - Señal de prueba (pulso rectangular).
    - Señal de audio real desde YouTube.

3. **Visualización completa**:
    - Señales en el tiempo: mensaje, señal modulada y señal recuperada.
    - Espectros en frecuencia.
    - Respuesta del filtro: diagrama de Bode y plano de polos y ceros.
""")

# Sidebar
st.sidebar.header("⚙️ Parámetros Generales")
fs = st.sidebar.slider("Frecuencia de muestreo (Hz)", 8000, 48000, 20000, 1000)
f_c = st.sidebar.slider("Frecuencia portadora (Hz)", 100, 10000, 2000, 100)
cutoff = st.sidebar.slider("Frecuencia de corte LPF (Hz)", 100, int(fs//2), 4000, 100)
url = st.sidebar.text_input("🔗 Enlace de YouTube para análisis de audio:")

# Funciones
def bode_and_zplane(b, a, fs):
    w, h = freqz(b, a, fs=fs)
    fig, ax = plt.subplots(2, 1, figsize=(8, 5))
    ax[0].plot(w, 20 * np.log10(np.abs(h)))
    ax[0].set_title("Bode - Magnitud [dB]"); ax[0].grid()
    ax[1].plot(w, np.unwrap(np.angle(h)))
    ax[1].set_title("Bode - Fase [rad]"); ax[1].grid()
    st.pyplot(fig)

    z, p, _ = tf2zpk(b, a)
    fig2, ax2 = plt.subplots()
    ax2.add_patch(plt.Circle((0, 0), 1, fill=False, linestyle='--'))
    ax2.plot(np.real(z), np.imag(z), 'o', label="Ceros")
    ax2.plot(np.real(p), np.imag(p), 'x', label="Polos")
    ax2.set_title("Plano de Polos y Ceros")
    ax2.axis("equal"); ax2.grid(); ax2.legend()
    st.pyplot(fig2)

def plot_spectrum(signal, fs, title):
    N = len(signal)
    f = fftshift(fftfreq(N, 1/fs))
    S = fftshift(np.abs(fft(signal)))
    fig, ax = plt.subplots(figsize=(10, 3))
    ax.plot(f, S)
    ax.set_title(title); ax.set_xlabel("Frecuencia [Hz]"); ax.grid()
    st.pyplot(fig)

# ===== 1. Pulso Rectangular =====
st.header("🔷 Pulso Rectangular")

duration = st.sidebar.slider("Duración del pulso (s)", 0.01, 0.5, 0.2, 0.01)
amplitude = st.sidebar.slider("Amplitud del pulso", 0.1, 5.0, 1.0, 0.1)

T = 1 / fs
t = np.arange(-0.5, 0.5, T)
m_t = amplitude * ((t >= -duration/2) & (t <= duration/2)).astype(float)
m_h = np.imag(hilbert(m_t))
carrier_cos = np.cos(2*np.pi*f_c*t)
carrier_sin = np.sin(2*np.pi*f_c*t)
ssb_signal = m_t * carrier_cos - m_h * carrier_sin

b, a = butter(5, cutoff, btype='low', fs=fs)
demod = ssb_signal * carrier_cos
rec_t = filtfilt(b, a, demod)

st.markdown("""
### 🧪 Sección 1: Pulso Rectangular

En esta sección, se genera una señal de prueba tipo **pulso rectangular** de duración y amplitud ajustables. Luego, se modula con SSB-AM y se recupera utilizando detección coherente y filtrado paso bajo.

- **$ m(t) $**: señal original.
- **SSB-AM**: señal modulada eliminando una banda lateral.
- **Mensaje recuperado**: señal demodulada tras pasar por un filtro pasa bajos.

Observe cómo el proceso conserva la forma de la señal original tras la demodulación.
""")

fig, ax = plt.subplots(3, 1, figsize=(10, 6), sharex=True)
ax[0].plot(t, m_t); ax[0].set_title("Mensaje m(t)"); ax[0].grid()
ax[1].plot(t, ssb_signal); ax[1].set_title("Señal SSB-AM"); ax[1].grid()
ax[2].plot(t, rec_t); ax[2].set_title("Mensaje recuperado"); ax[2].grid()
st.pyplot(fig)

plot_spectrum(ssb_signal, fs, "Espectro de la señal SSB-AM (Pulso)")
bode_and_zplane(b, a, fs)

# ===== 2. Audio desde YouTube =====
st.header("🔊 Audio desde YouTube")

audio_loaded = False
if url:
    try:
        with st.spinner("🔽 Descargando y convirtiendo audio..."):
            ydl_opts = {'format': 'bestaudio', 'outtmpl': 'temp_audio.%(ext)s'}
            with youtube_dl.YoutubeDL(ydl_opts) as ydl:
                ydl.download([url])

            subprocess.run(['ffmpeg', '-y', '-i', 'temp_audio.webm',
                            '-ar', str(fs), '-ac', '1', 'audio.wav'],
                           stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

            sr, audio = wavfile.read('audio.wav')
            audio = audio / np.max(np.abs(audio))
            audio = audio[:int(fs*5)]
            t_audio = np.linspace(0, len(audio)/fs, len(audio), endpoint=False)
            audio_loaded = True
    except Exception as e:
        st.error(f"Error descargando audio: {e}")

if audio_loaded:
    audio_h = np.imag(hilbert(audio))
    ssb_audio = audio * np.cos(2*np.pi*f_c*t_audio) - audio_h * np.sin(2*np.pi*f_c*t_audio)
    demod_audio = ssb_audio * np.cos(2*np.pi*f_c*t_audio)
    rec_audio = filtfilt(b, a, demod_audio)

    st.markdown("### 🎧 Reproducción de audio")
    st.audio(audio, format="audio/wav", sample_rate=fs)
    st.audio(rec_audio, format="audio/wav", sample_rate=fs)

    st.markdown("""
### 🎵 Sección 2: Señal Real (Audio de YouTube)

Aquí el usuario puede ingresar un enlace a un video de YouTube. El sistema descarga el audio, lo normaliza, y realiza el mismo proceso de modulación/demodulación SSB-AM.

Esto permite:
- Escuchar la calidad de la señal original vs la recuperada.
- Visualizar su representación temporal y espectral.
- Ver cómo responde el filtro pasa bajos (diseñado con Butterworth IIR) que recupera el mensaje.

Este tipo de análisis es muy útil en contextos reales de telecomunicaciones, donde se desea transmitir solo una banda lateral para ahorrar espectro.
""")

    fig3, axs = plt.subplots(3, 1, figsize=(10, 6), sharex=True)
    axs[0].plot(t_audio, audio); axs[0].set_title("Audio original"); axs[0].grid()
    axs[1].plot(t_audio, ssb_audio); axs[1].set_title("SSB-AM (modulada)"); axs[1].grid()
    axs[2].plot(t_audio, rec_audio); axs[2].set_title("Audio recuperado"); axs[2].grid()
    st.pyplot(fig3)

    plot_spectrum(ssb_audio, fs, "Espectro de SSB-AM (audio)")
    bode_and_zplane(b, a, fs)

    st.markdown("""
### 📊 Comparación Temporal

Esta gráfica final superpone la señal original y la señal recuperada para evaluar la fidelidad del sistema de modulación y demodulación SSB-AM.

Idealmente, ambas señales deberían ser casi indistinguibles (exceptuando posibles efectos de fase o ganancia introducidos por el filtro).
""")

    fig4, axc = plt.subplots(figsize=(10, 3))
    axc.plot(t_audio[:5000], audio[:5000], label="Original")
    axc.plot(t_audio[:5000], rec_audio[:5000], '--', label="Recuperado")
    axc.set_title("Comparación temporal (Audio original vs recuperado)")
    axc.set_xlabel("Tiempo [s]")
    axc.legend(); axc.grid()
    st.pyplot(fig4)
```
